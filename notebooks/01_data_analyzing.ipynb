## Data quality check (olist_products_dataset)

#### Checking for empty values

```python
df_products.isna().sum()
```

```python
df_products['product_category_name'].isna().mean()
```

#### The presence of gaps in the columns 'product_name_length', 'product_description_length', 'product_photos_qty', 'product_weight_g', 'product_length_cm', 'product_height_cm', and 'product_width_cm' are not significant at the stage of analyzing the company's sales. However, gaps in the 'product_category_name' column are problematic, since it was planned to use this column to analyze the popularity of categories by region, time, etc. The proportion of missing data in the entire sample is 1,85%.

#### Assessing the impact of missing data

```python
df_merged = (df_order_items.merge(df_products, on='product_id'))
revenue_null_product_category_name = df_merged[df_merged['product_category_name'].isna()]['price'].sum() + df_merged[df_merged['product_category_name'].isna()]['freight_value'].sum()
revenue_all_order_items = df_order_items['price'].sum() + df_order_items['freight_value'].sum()
revenue_null_product_category_name / revenue_all_order_items
```

#### The revenue share of products without a category name is 1,31%, which cannot significantly distort the data. For such products, the 'Unknown' category will be entered in the data_mart table.
